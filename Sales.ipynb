# 📊 Retail Sales Analysis – Superstore Dataset

## **Introduction**

This project focuses on analyzing retail sales data from a Superstore business. The dataset contains information about orders, customers, products, sales, profits, discounts, and shipping details.

The objective of this analysis is to identify key business drivers, evaluate profitability across different dimensions, understand customer behavior, and uncover opportunities to improve overall business performance.

#### Metadata
- **Row ID**: Unique ID for each row.
- **Order ID**: Unique Order ID for each Customer.
- **Order Date**: Order Date of the product.
- **Ship Date**: Shipping Date of the Product.
- **Ship Mode**: Shipping Mode specified by the Customer.
- **Customer ID**: Unique ID to identify each Customer.
- **Customer Name**: Name of the Customer.
- **Segment**: The segment where the Customer belongs.
- **Country**: Country of residence of the Customer.
- **City**: City of residence of the Customer.
- **State**: State of residence of the Customer.
- **Postal Code**: Postal Code of every Customer.
- **Region**: Region where the Customer belongs.
- **Product ID**: Unique ID of the Product.
- **Category**: Category of the product ordered.
- **Sub-Category**: Sub-Category of the product ordered.
- **Product Name**: Name of the Product.
- **Sales**: Sales of the Product.
- **Quantity**: Quantity of the Product.
- **Discount**: Discount provided.
- **Profit**: Profit/Loss incurred.

The analysis covers several important areas:

* Sales Performance Analysis
* Profitability Analysis
* Customer Analysis
* Discount Impact Analysis
* Regional Performance Analysis
* Time-Based Sales Trends

Through these analyses, the project aims to generate actionable business insights that can support data-driven decision making and improve profitability.

## 🏷 Discount Analysis


```python
df['Discount Bin'] = pd.cut(df['Discount'],bins=5)
df_discount = df.groupby('Discount Bin')['Sales'].mean().reset_index() 

plt.figure(figsize=(8,5))
plt.scatter(df_discount['Discount Bin'].astype(str),df_discount['Sales'] , marker='o')

plt.xlabel('Discount Range')
plt.ylabel('Average Sales')
plt.title('Average Sales by Discount Level')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
df['Discount Bin'] = pd.cut(df['Discount'],bins=5)
df_discount = df.groupby('Discount Bin')['Profit'].mean().reset_index() 

plt.figure(figsize=(8,5))
plt.scatter(df_discount['Discount Bin'].astype(str),df_discount['Profit'] , marker='o')

plt.xlabel('Discount Range')
plt.ylabel('Average Profit')
plt.title('Average Profit by Discount Level')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
df['Discount Bin'] = pd.cut(df['Discount'],bins=5)
df_discount = df.groupby('Discount Bin')['Profit_Margin'].mean().reset_index() 

plt.figure(figsize=(8,5))
plt.scatter(df_discount['Discount Bin'].astype(str),df_discount['Profit_Margin'] , marker='o')

plt.xlabel('Discount Range')
plt.ylabel('Average Profit_Margin')
plt.title('Average Profit_Margin by Discount Level')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

**Key Findings**

* Average profit margin decreases consistently as discount levels increase.
* Discounts below 32% generate positive profit and healthy profit margins.
* The highest average sales occur in the 32%–48% discount range.
* Despite higher sales, discounts above 32% result in negative average profit and profit margins.
* Very high discounts (above 48%) are associated with significant financial losses.
* The 64%–80% discount range has the lowest sales performance while also producing the largest losses.
* Higher discounts do not necessarily lead to higher sales and can severely damage profitability.

**Recommendations**

* Limit the use of discounts above 32%, as they are generally unprofitable.
* Focus promotional strategies on low to moderate discount levels where profitability remains positive.
* Review products receiving discounts above 48% and identify the causes of recurring losses.
* Evaluate discount policies at the Category and Sub-Category level to identify products that can sustain promotions profitably.
* Consider alternative marketing strategies instead of deep discounts to increase sales.
* Monitor both sales growth and profit margin simultaneously when designing future discount campaigns.

```python
df = df.sort_values(['Customer Name', 'Order Date'])

df['Days_Between'] = (
    df.groupby('Customer Name')['Order Date']
      .diff()
      .dt.days
)


customer_stats = df.groupby('Customer Name').agg(
    First_Order=('Order Date', 'min'),
    Last_Order=('Order Date', 'max'),
    Orders=('Order ID', 'nunique'),
    Sales=('Sales', 'sum'),
    Profit=('Profit', 'sum'),
    Avg_Days_Between=('Days_Between', 'mean')
)

customer_stats['Lifetime_Days'] = (
    customer_stats['Last_Order']
    - customer_stats['First_Order']
).dt.days

customer_stats = customer_stats.dropna(
    subset=['Avg_Days_Between']
)


plt.figure(figsize=(12,8))

scatter = plt.scatter(
    customer_stats['Lifetime_Days'],
    customer_stats['Avg_Days_Between'],
    s=customer_stats['Sales']/50,   
    alpha=0.6
)

plt.xlabel('Customer Lifetime (Days)')
plt.ylabel('Average Days Between Orders')
plt.title('Customer Loyalty Analysis')

plt.grid(alpha=0.3)
plt.show()
```

* Most customers have a lifetime between 1,000 and 1,400 days, indicating strong long-term retention.
* The majority of active customers place orders every 50–120 days, forming the company's core customer base.
* A group of highly valuable customers combines long customer lifetime, frequent purchases, and high sales volume.
* Several customers show very long intervals between purchases, suggesting a higher risk of customer churn.
* Customer lifetime alone is not a sufficient indicator of loyalty; purchase frequency must also be considered when evaluating customer engagement
